```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from anthropic import Anthropic

client = Anthropic()
```

```python
model = "claude-sonnet-4-6"
messages = []

system_message = "You are a python programming languge expert. Give just a precise code snippet without explanation."

def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text
    }
    messages.append(user_message)

def chat(messages, system_message=None):
    params = {
        "model": model,
        "messages": messages,
        "max_tokens": 1024,
    }
    if system_message:
        params["system"] = system_message

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
add_user_message(messages, "Write a python function that checks a string for duplicate words.")
answer = chat(messages, system_message)
print(answer)
```
