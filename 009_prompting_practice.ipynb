```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()
client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages: list, text: str):
    message = {
        "role": "user",
        "content": text
    }
    messages.append(message)


def chat(messages: list, system_prompt=None):
    props = {
        "model": model,
        "max_tokens": 1024,
        "messages": messages
    }
    if system_prompt:
        props["system"] = system_prompt

    response = client.messages.create(**props)
    return response.content[0].text

```

```python
messages = []
add_user_message(messages, "How many calories does a pound of steamed Broccili contain?")
answer = chat(messages, "You are a dietitian. Give short and precise answers.")

print(answer)
```
