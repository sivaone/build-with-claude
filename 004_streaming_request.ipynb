```python
from dotenv import load_dotenv
from anthropic import Anthropic
load_dotenv()

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Code to get Raw events
response_stream = client.messages.create(
    model=model,
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": "Write a small paragraph about fake futuristic city"
        }
    ],
    stream=True
)

for response in response_stream:
    print(response)
```

```python
# Easy to use sdk to get text from stream
with client.messages.stream(
    model=model,
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": "Write a small paragraph about fake futuristic city"
        }
    ]
) as stream:
    for text in stream.text_stream:
        print(text, end="")
```
